# Автоматический подбор музыки для видео



```
import base64, io, os, pathlib, subprocess, sys, zipfile
PROJECT = pathlib.Path('/content/music-video-retrieval')
PROJECT.mkdir(exist_ok=True)
zipfile.ZipFile(io.BytesIO(base64.b64decode(''))).extractall(PROJECT)
os.chdir(PROJECT)
```

## Зависимости


```
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
```

## Проверки


```
subprocess.run([sys.executable, '-c', "import torch; assert torch.cuda.is_available(); print(torch.cuda.get_device_name(0))"], check=True)
for command in ([sys.executable, '-m', 'unittest', 'discover', '-s', 'tests', '-v'], [sys.executable, '-m', 'src.revision.smoke']):
    result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    print(result.stdout)
    result.check_returncode()
```

## Подготовка данных и обучение



```
START = 'prepare'
log_dir = PROJECT / 'runs/revision'
log_dir.mkdir(parents=True, exist_ok=True)
with (log_dir / 'pipeline.log').open('a') as log:
    process = subprocess.Popen([sys.executable, '-u', '-m', 'src.revision.run', '--start', START],
                               stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in process.stdout:
        print(line, end='')
        log.write(line)
        log.flush()
    exit_code = process.wait()
print('Exit code:', exit_code)
if exit_code != 0:
    raise RuntimeError(f"Pipeline failed with exit code {exit_code}; inspect status.json")
```

## Экспорт результатов


```
archive_path = pathlib.Path('/content/music_video_results.zip')
with zipfile.ZipFile(archive_path, 'w', zipfile.ZIP_DEFLATED) as archive:
    for folder in ('runs/revision', 'data/revision/features'):
        for path in (PROJECT / folder).rglob('*'):
            if path.is_file(): archive.write(path, str(path.relative_to(PROJECT)))
    for path in (PROJECT / 'data/revision').glob('*.csv'):
        archive.write(path, str(path.relative_to(PROJECT)))
    archive.write(PROJECT / 'configs/revision.json', 'configs/revision.json')
from google.colab import files
files.download(str(archive_path))
```
